```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
from wordcloud import WordCloud
```

```python
df = pd.read_csv('modeling_results.csv')

features_to_correlate = [
    'word_count', 'danceability', 'energy', 
    'valence', 'loudness', 'speechiness', 'tempo',
    'Topic_0_Prob', 'Topic_1_Prob', 'Topic_2_Prob'
]

corr_matrix = df[features_to_correlate].corr(method='spearman')
```

```python
out_dir = os.path.join('..', 'result', 'figure','1_text_length')
os.makedirs(out_dir, exist_ok=True)

sns.set_theme(style="whitegrid", palette="muted")

plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='word_count', bins=40, kde=True, color='#4CB391')
plt.title('Distribution of Lyrics Word Count', fontsize=16, fontweight='bold', pad=15)
plt.xlabel('Word Count (Total Words)', fontsize=12)
plt.ylabel('Number of Tracks', fontsize=12)

median_length = df['word_count'].median()
plt.axvline(median_length, color='red', linestyle='--', label=f'Median: {median_length:.0f}')
plt.legend()
plt.tight_layout()
p1 = os.path.join(out_dir, '1A_WordCount_Histogram.png')
plt.savefig(p1, dpi=300)
plt.close()

# ================= Figure 1-B：Histogram =================
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='lexical_richness', bins=40, kde=True, color='#8993F8')
plt.title('Distribution of Lexical Richness', fontsize=16, fontweight='bold', pad=15)
plt.xlabel('Lexical Richness (Unique Words / Total Words)', fontsize=12)
plt.ylabel('Number of Tracks', fontsize=12)
median_richness = df['lexical_richness'].median()
plt.axvline(median_richness, color='red', linestyle='--', label=f'Median: {median_richness:.2f}')
plt.legend()
plt.tight_layout()
p2 = os.path.join(out_dir, '1B_LexicalRichness_Histogram.png')
plt.savefig(p2, dpi=300)
plt.close()

# ================= Figure 1-C：Boxplot =================
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x='Dominant_Topic', y='word_count', palette='Set2')
sns.stripplot(data=df, x='Dominant_Topic', y='word_count', color=".25", alpha=0.3, jitter=True)
plt.title('Word Count Distribution across LDA Topics', fontsize=16, fontweight='bold', pad=15)
plt.xlabel('LDA Topic Number', fontsize=12)
plt.ylabel('Word Count', fontsize=12)
plt.tight_layout()
p3 = os.path.join(out_dir, '1C_Topic_vs_WordCount_Boxplot.png')
plt.savefig(p3, dpi=300)
plt.close()

# ================= Figure 1-D：lexical_richness (Boxplot) =================
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x='Dominant_Topic', y='lexical_richness', palette='Set3')
sns.stripplot(data=df, x='Dominant_Topic', y='lexical_richness', color=".25", alpha=0.3, jitter=True)
plt.title('Lexical Richness across LDA Topics', fontsize=16, fontweight='bold', pad=15)
plt.xlabel('LDA Topic Number', fontsize=12)
plt.ylabel('Lexical Richness', fontsize=12)
plt.tight_layout()
p4 = os.path.join(out_dir, '1D_Topic_vs_Richness_Boxplot.png')
plt.savefig(p4, dpi=300)
plt.close()
```

```python
out_dir = os.path.join('..', 'result', 'figure','2_correlation')
os.makedirs(out_dir, exist_ok=True)


pure_nlp_features = [
    'word_count', 'lexical_richness', 'avg_word_length', 'text_sentiment',
    'Topic_0_Prob', 'Topic_1_Prob', 'Topic_2_Prob', 'Topic_3_Prob', 'Topic_4_Prob'
]


corr_matrix = df[pure_nlp_features].corr(method='spearman')


sns.set_theme(style="white")
plt.figure(figsize=(11, 9))


sns.heatmap(corr_matrix, 
            annot=True, 
            fmt=".2f", 
            cmap="PRGn", 
            center=0, 
            square=True, 
            linewidths=.5, 
            cbar_kws={"shrink": .8})

plt.title('Correlation Heatmap of Pure NLP Features', fontsize=16, fontweight='bold', pad=20)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()

p_heatmap = os.path.join(out_dir, '2A_NLP_Correlation_Heatmap.png')
plt.savefig(p_heatmap, dpi=300)
plt.close()
```

```python
global_tfidf_file = 'global_tfidf_ranking.csv'
model_file = 'modeling_results.csv'

df_global = pd.read_csv(global_tfidf_file)
df_main = pd.read_csv(model_file)
df_main['processed_lyrics'] = df_main['processed_lyrics'].fillna('')

out_dir = os.path.join('..', 'result', 'figure','3_tfidf')
os.makedirs(out_dir, exist_ok=True)

sns.set_theme(style="whitegrid")
plt.figure(figsize=(12, 8))

top_20_words = df_global.head(20)

sns.barplot(data=top_20_words, x='TF-IDF_score', y='term', palette='viridis')
plt.title('Top 20 Global Keywords by TF-IDF Score', fontsize=18, fontweight='bold', pad=15)
plt.xlabel('Cumulative TF-IDF Score', fontsize=14)
plt.ylabel('Keywords', fontsize=14)
plt.tight_layout()

p1 = os.path.join(out_dir, '3A_Top20_TFIDF_BarChart.png')
plt.savefig(p1, dpi=300)
plt.close()


tfidf_dict = dict(zip(df_global['term'], df_global['TF-IDF_score']))

wordcloud_global = WordCloud(width=1200, height=800, 
                             background_color='white', 
                             colormap='magma',
                             max_words=150).generate_from_frequencies(tfidf_dict)

plt.figure(figsize=(12, 8))
plt.imshow(wordcloud_global, interpolation='bilinear')
plt.axis('off') 
plt.title('Global Lyrics Word Cloud (Sized by TF-IDF)', fontsize=18, fontweight='bold', pad=20)
plt.tight_layout()

p2 = os.path.join(out_dir, '3B_Global_WordCloud.png')
plt.savefig(p2, dpi=300)
plt.close()


colormaps = ['Blues', 'Reds', 'Greens', 'Purples', 'Oranges']

for topic_idx in sorted(df_main['Dominant_Topic'].unique()):
    topic_lyrics = " ".join(df_main[df_main['Dominant_Topic'] == topic_idx]['processed_lyrics'].tolist())
    
    if len(topic_lyrics.strip()) == 0:
        continue

    wc_topic = WordCloud(width=800, height=600, 
                         background_color='white', 
                         colormap=colormaps[topic_idx % len(colormaps)],
                         max_words=100, 
                         collocations=False).generate(topic_lyrics)
    
    plt.figure(figsize=(8, 6))
    plt.imshow(wc_topic, interpolation='bilinear')
    plt.axis('off')
    plt.title(f'Word Cloud for LDA Topic {topic_idx}', fontsize=16, fontweight='bold', pad=15)
    plt.tight_layout()
    
    p_topic = os.path.join(out_dir, f'3_Topic_{topic_idx}_WordCloud.png')
    plt.savefig(p_topic, dpi=300)
    plt.close()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os


model_file = 'modeling_results.csv'
df = pd.read_csv(model_file)
df['processed_lyrics'] = df['processed_lyrics'].fillna('')


df['fetched_at'] = pd.to_datetime(df['fetched_at'], errors='coerce')

df = df.dropna(subset=['fetched_at']) 

df['year_month'] = df['fetched_at'].dt.to_period('M').astype(str)
df = df.sort_values('year_month')

out_dir = os.path.join('..', 'result', 'figure','4_timeseries')
os.makedirs(out_dir, exist_ok=True)


topic_cols = ['Topic_0_Prob', 'Topic_1_Prob', 'Topic_2_Prob', 'Topic_3_Prob', 'Topic_4_Prob']
monthly_topics = df.groupby('year_month')[topic_cols].mean()

plt.figure(figsize=(12, 6))
monthly_topics.plot(kind='area', stacked=True, alpha=0.75, colormap='Set3', figsize=(12, 6))

plt.title('Evolution of LDA Topics Over Time', fontsize=16, fontweight='bold', pad=15)
plt.xlabel('Time (Year-Month)', fontsize=12)
plt.ylabel('Average Topic Probability', fontsize=12)
plt.legend(title='LDA Topics', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xticks(rotation=45)
plt.tight_layout()

p1 = os.path.join(out_dir, '4A_Topic_Evolution_AreaChart.png')
plt.savefig(p1, dpi=300)
plt.close()



sns.set_theme(style="whitegrid")

global_tfidf_file = 'global_tfidf_ranking.csv'
df_global = pd.read_csv(global_tfidf_file)
top_3_words = df_global['term'].head(3).tolist()

for word in top_3_words:
    df[f'contains_{word}'] = df['processed_lyrics'].str.contains(rf'\b{word}\b', regex=True).astype(int)

word_cols = [f'contains_{word}' for word in top_3_words]
monthly_words = df.groupby('year_month')[word_cols].mean()

plt.figure(figsize=(12, 6))
markers = ['o', 's', '^']
colors = ['#e63946', '#457b9d', '#2a9d8f']

for i, word in enumerate(top_3_words):
    plt.plot(monthly_words.index, monthly_words[f'contains_{word}'], 
             marker=markers[i], color=colors[i], linewidth=2.5, markersize=8, 
             label=f'Keyword: "{word}"')

plt.title('Top 3 Global Keywords Popularity Over Time', fontsize=16, fontweight='bold', pad=15)
plt.xlabel('Time (Year-Month)', fontsize=12)
plt.ylabel('Proportion of Tracks Containing Keyword', fontsize=12)
plt.xticks(rotation=45)
plt.legend(title='Keywords', fontsize=11)
plt.tight_layout()

p2 = os.path.join(out_dir, '4B_Keyword_Trend_LineChart.png')
plt.savefig(p2, dpi=300)
plt.close()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
from scipy.stats import zscore
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import PCA

model_file = 'modeling_results.csv'
df = pd.read_csv(model_file)
df['processed_lyrics'] = df['processed_lyrics'].fillna('')

out_dir = os.path.join('..', 'result', 'figure','5_anomalies')
os.makedirs(out_dir, exist_ok=True)


df['z_word_count'] = zscore(df['word_count'])
df['z_lexical_richness'] = zscore(df['lexical_richness'])

df['is_outlier'] = (df['z_word_count'].abs() > 2.5) | (df['z_lexical_richness'].abs() > 2.5)

plt.figure(figsize=(10, 8))
sns.scatterplot(data=df[~df['is_outlier']], x='z_word_count', y='z_lexical_richness', 
                color='lightgrey', alpha=0.6, label='Normal Tracks')
sns.scatterplot(data=df[df['is_outlier']], x='z_word_count', y='z_lexical_richness', 
                color='red', edgecolor='black', s=80, label='Anomalies (Extreme Z-Score)')

extreme_outliers = df[df['is_outlier']].copy()
extreme_outliers['total_deviation'] = extreme_outliers['z_word_count'].abs() + extreme_outliers['z_lexical_richness'].abs()
top_extremes = extreme_outliers.sort_values('total_deviation', ascending=False).head(5)

for _, row in top_extremes.iterrows():
    plt.text(row['z_word_count'] + 0.1, row['z_lexical_richness'] + 0.1, 
             row['track_name'][:15] + '...', fontsize=9, fontweight='bold', color='darkred')

plt.axvline(0, color='blue', linestyle='--', alpha=0.3)
plt.axhline(0, color='blue', linestyle='--', alpha=0.3)

plt.title('Z-Score Outlier Detection: Word Count vs. Lexical Richness', fontsize=16, fontweight='bold', pad=15)
plt.xlabel('Z-Score of Word Count', fontsize=12)
plt.ylabel('Z-Score of Lexical Richness', fontsize=12)
plt.legend()
plt.tight_layout()

p1 = os.path.join(out_dir, '5A_ZScore_Outliers_Scatter.png')
plt.savefig(p1, dpi=300)
plt.close()


tfidf_vectorizer = TfidfVectorizer(max_df=0.9, min_df=5)
tfidf_matrix = tfidf_vectorizer.fit_transform(df['processed_lyrics']).toarray()

pca = PCA(n_components=2, random_state=42)
pca_result = pca.fit_transform(tfidf_matrix)

df['pca_x'] = pca_result[:, 0]
df['pca_y'] = pca_result[:, 1]

plt.figure(figsize=(11, 8))

sns.scatterplot(data=df, x='pca_x', y='pca_y', hue='Dominant_Topic', 
                palette='Set1', alpha=0.7, s=50, edgecolor='white')

plt.title('2D PCA Projection of TF-IDF Space (Colored by LDA Topic)', fontsize=16, fontweight='bold', pad=15)
plt.xlabel('Principal Component 1 (Semantic Variance)', fontsize=12)
plt.ylabel('Principal Component 2 (Semantic Variance)', fontsize=12)
plt.legend(title='LDA Topic', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()

p2 = os.path.join(out_dir, '5B_TFIDF_PCA_Clustering.png')
plt.savefig(p2, dpi=300)
plt.close()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
from sklearn.feature_extraction.text import CountVectorizer

model_file = 'modeling_results.csv'
df = pd.read_csv(model_file)
df['processed_lyrics'] = df['processed_lyrics'].fillna('')

out_dir = os.path.join('..', 'result', 'figure','6_lda_topics')
os.makedirs(out_dir, exist_ok=True)

plt.figure(figsize=(10, 8))

topic_counts = df['Dominant_Topic'].value_counts().sort_index()


colors = sns.color_palette("pastel")[0:len(topic_counts)]
explode = [0.05 if i == topic_counts.idxmax() else 0 for i in range(len(topic_counts))]

plt.pie(topic_counts, 
        labels=[f"Topic {i}" for i in topic_counts.index], 
        colors=colors, 
        explode=explode,
        autopct='%1.1f%%',   
        startangle=140, 
        textprops={'fontsize': 14, 'fontweight': 'bold'},
        shadow=True)

plt.title('Distribution of Dominant LDA Topics across the Corpus', fontsize=18, fontweight='bold', pad=20)
plt.tight_layout()

p_pie = os.path.join(out_dir, '6A_Dominant_Topic_PieChart.png')
plt.savefig(p_pie, dpi=300)
plt.close()



sns.set_theme(style="whitegrid")
colormaps = ['Blues_r', 'Reds_r', 'Greens_r', 'Purples_r', 'Oranges_r']

for topic_idx in sorted(df['Dominant_Topic'].unique()):

    topic_docs = df[df['Dominant_Topic'] == topic_idx]['processed_lyrics'].tolist()
    
    if not topic_docs:
        continue
        
    vectorizer = CountVectorizer(max_df=0.9, min_df=2)
    tf_matrix = vectorizer.fit_transform(topic_docs)
    
    words = vectorizer.get_feature_names_out()
    word_counts = tf_matrix.sum(axis=0).A1
    
    topic_df = pd.DataFrame({'word': words, 'count': word_counts})
    top_10_words = topic_df.sort_values(by='count', ascending=False).head(10)
    
    plt.figure(figsize=(10, 6))
    sns.barplot(data=top_10_words, x='count', y='word', palette=colormaps[topic_idx % len(colormaps)])
    
    plt.title(f'Top 10 Representative Words for LDA Topic {topic_idx}', fontsize=16, fontweight='bold', pad=15)
    plt.xlabel('Word Frequency within Topic', fontsize=12)
    plt.ylabel('Keywords', fontsize=12)
    plt.tight_layout()
    
    p_bar = os.path.join(out_dir, f'6B_Topic_{topic_idx}_TopWords_Bar.png')
    plt.savefig(p_bar, dpi=300)
    plt.close()
```

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import LatentDirichletAllocation
import pyLDAvis

out_dir = os.path.join('..', 'result', 'figure')
os.makedirs(out_dir, exist_ok=True)

model_file = 'modeling_results.csv'
df = pd.read_csv(model_file)
docs = df['processed_lyrics'].fillna('').tolist()

tf_vectorizer = CountVectorizer(max_df=0.9, min_df=5)
tf_matrix = tf_vectorizer.fit_transform(docs)

N_TOPICS = 5 
lda_model = LatentDirichletAllocation(n_components=N_TOPICS, max_iter=10, random_state=42)
lda_model.fit(tf_matrix)


try:
    import pyLDAvis.lda_model
    vis_data = pyLDAvis.lda_model.prepare(lda_model, tf_matrix, tf_vectorizer)
except ImportError:
    import pyLDAvis.sklearn
    vis_data = pyLDAvis.sklearn.prepare(lda_model, tf_matrix, tf_vectorizer)


output_html = "LDA_Interactive_Dashboard.html"
output_path = os.path.join(out_dir, output_html)
pyLDAvis.save_html(vis_data, output_path)
```
